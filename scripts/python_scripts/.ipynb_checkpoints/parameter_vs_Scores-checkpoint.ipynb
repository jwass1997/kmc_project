```python
def plot_topK_params_vs_logmse(params_topk, mse_topk, param_names=None):
    P = np.asarray(params_topk)          # (K,6)
    y = np.asarray(mse_topk).reshape(-1) # (K,)
    K, D = P.shape
    assert D == 6

    if param_names is None:
        param_names = [f"p{i}" for i in range(D)]

    order = np.argsort(y)   # best first
    P = P[order]
    y = y[order]
    ly = np.log10(y)
    vmin, vmax = np.min(ly), np.max(ly)

    fig, axes = plt.subplots(2, 3, figsize=(10, 6), sharey=True)
    axes = axes.ravel()

    sc = None
    for i in range(D):
        ax = axes[i]
        ax.minorticks_on()
        ax.tick_params(top=True, right=True, which='both')
        ax.tick_params(direction='in', which='both')
        sc = ax.scatter(P[:, i], ly, c=ly, cmap='viridis', s=40, alpha=0.65, linewidths=0)

        # Optional: connect points in rank order to see drift along the valley
        #ax.plot(P[:, i], ly, linewidth=0.6, alpha=0.5)

        ax.set_xlabel(param_names[i])
        ax.grid(True, alpha=0.5, lw=0.5)

    if sc is not None:
        cb = fig.colorbar(sc, ax=axes, fraction=0.02, pad=0.01)
        cb.set_label("$\log_{10}(\mathcal{E}_{\mathrm{aff}})$")
        
    axes[0].set_ylabel("$\log_{10}(\mathcal{E}_{\mathrm{aff}})$")
    axes[3].set_ylabel("$\log_{10}(\mathcal{E}_{\mathrm{aff}})$")
    hs = 0.25
    fig.subplots_adjust(hspace=hs)
    #fig.tight_layout()
    x0, y0, w, h = axes[-1].get_position().bounds
    cb.ax.set_position([x0+0.02+w, y0, w, h])

    return fig, axes

import numpy as np
import matplotlib.pyplot as plt

def corner_topk_2d(params_topk, mse_topk, param_names=None, cmap="jet_r"):
    """
    params_topk: (K, D) array-like (D=6)
    mse_topk:    (K,)   array-like
    """
    P = np.asarray(params_topk)
    y = np.asarray(mse_topk).reshape(-1)
    K, D = P.shape
    if y.shape[0] != K:
        raise ValueError("mse_topk must have same length as params_topk rows")

    if param_names is None:
        param_names = [f"p{i}" for i in range(D)]

    # sort by score so "best" points are drawn last (on top)
    order = np.argsort(y)
    P = P[order]
    y = y[order]
    c = np.log10(y)

    fig, axes = plt.subplots(D, D, figsize=(2.1*D, 2.1*D), squeeze=False)
    vmin, vmax = np.min(c), np.max(c)

    sc = None
    for i in range(D):
        for j in range(D):
            ax = axes[i, j]
            ax.minorticks_on()
            ax.tick_params(top=True, right=True, which='both')
            ax.tick_params(direction='in', which='both')

            if i == j:
                ax.hist(P[:, j], bins=15)
            elif i > j:
                sc = ax.scatter(
                    P[:, j], P[:, i],
                    c=c, cmap=cmap, vmin=vmin, vmax=vmax,
                    s=32, alpha=0.6, linewidths=0
                )
            else:
                ax.axis("off")

            # tidy labels
            if i == D - 1 and j <= i:
                ax.set_xlabel(param_names[j])
            else:
                ax.set_xticklabels([])
            if j == 0 and i >= j:
                ax.set_ylabel(param_names[i])
            else:
                ax.set_yticklabels([])

            ax.grid(True, alpha=0.15)
    
    if sc is not None:
        cb = fig.colorbar(sc, ax=axes, fraction=0.02, pad=0.01)
        cb.set_label("$\log_{10}(\mathcal{E}_{\mathrm{aff}})$")
    axes[0, 0].set_yticklabels([])
    x0, y0, w, h = axes[-1, -1].get_position().bounds
    print(h)
    cb.ax.set_position([x0+0.02+w, y0, w, 2*h+0.022])
    return fig, axes

import numpy as np
import matplotlib.pyplot as plt

def corner_topk_2d_hex(params_topk, mse_topk, param_names=None, cmap="jet_r",
                  gridsize=40, bins_diag=20, overlay_best=80,
                  reduce_fn="median", density_bins="log"):
    """
    Corner-style plot for Top-K designs (K x D), robust to overplotting via hexbin.
    
    Parameters
    ----------
    params_topk : array-like, shape (K, D)
        Top-K parameter vectors.
    mse_topk : array-like, shape (K,)
        Score to color by (MSE). Lower is better.
    param_names : list of str, optional
        Names for parameters.
    cmap : str
        Matplotlib colormap for MSE coloring.
    gridsize : int
        Hexbin resolution for off-diagonal plots.
    bins_diag : int
        Histogram bins on diagonal.
    overlay_best : int
        Overlay the best-N points (scatter) on top of hexbin density.
        Set 0 to disable.
    reduce_fn : {"median","mean","min"}
        How to aggregate log10(MSE) within each hex cell for the colored hexbin.
    density_bins : None or "log"
        If not None, also draw a background density hexbin using counts (optionally log-binned).
    
    Returns
    -------
    fig, axes
    """
    P = np.asarray(params_topk)
    y = np.asarray(mse_topk).reshape(-1)

    if P.ndim != 2:
        raise ValueError("params_topk must be a 2D array (K, D)")
    K, D = P.shape
    if y.shape[0] != K:
        raise ValueError("mse_topk must have same length as params_topk rows")

    if param_names is None:
        param_names = [f"p{i}" for i in range(D)]
    if len(param_names) != D:
        raise ValueError("param_names must have length D")

    # MSE: smaller is better
    # For overlay: best points are the smallest mse
    best_idx = np.argsort(y)[:min(int(overlay_best), K)] if overlay_best else np.array([], dtype=int)

    # log10 color values (avoid log of 0)
    eps = np.finfo(float).tiny
    c = np.log10(np.maximum(y, eps))
    vmin, vmax = np.min(c), np.max(c)

    reduce_map = {
        "median": np.median,
        "mean": np.mean,
        "min": np.min
    }
    if reduce_fn not in reduce_map:
        raise ValueError("reduce_fn must be one of: 'median', 'mean', 'min'")
    reduce_C = reduce_map[reduce_fn]

    fig, axes = plt.subplots(D, D, figsize=(2.2 * D, 2.2 * D), squeeze=False)
    sc_for_cb = None

    for i in range(D):
        for j in range(D):
            ax = axes[i, j]
            ax.minorticks_on()
            ax.tick_params(top=True, right=True, which='both')
            ax.tick_params(direction='in', which='both')
            ax.grid(True, alpha=0.15)

            if i == j:
                ax.hist(P[:, j], bins=bins_diag)
            elif i > j:
                # optional density background (counts)
                if density_bins is not None:
                    ax.hexbin(
                        P[:, j], P[:, i],
                        gridsize=gridsize,
                        mincnt=1,
                        bins=density_bins,
                        linewidths=0.0
                    )

                # colored hexbin: aggregate log10(MSE) in each cell
                hb = ax.hexbin(
                    P[:, j], P[:, i],
                    C=c,
                    reduce_C_function=reduce_C,
                    gridsize=gridsize,
                    mincnt=1,
                    cmap=cmap,
                    vmin=vmin, vmax=vmax,
                    linewidths=0.0
                )
                sc_for_cb = hb

                # overlay best points for clarity
                if best_idx.size:
                    ax.scatter(
                        P[best_idx, j], P[best_idx, i],
                        c=c[best_idx], cmap=cmap, vmin=vmin, vmax=vmax,
                        s=26, alpha=0.95, linewidths=0
                    )
            else:
                ax.axis("off")

            # tidy labels
            if i == D - 1 and j <= i:
                ax.set_xlabel(param_names[j])
            else:
                ax.set_xticklabels([])
            if j == 0 and i >= j:
                ax.set_ylabel(param_names[i])
            else:
                ax.set_yticklabels([])

    # colorbar from the colored-hexbin
    if sc_for_cb is not None:
        cb = fig.colorbar(sc_for_cb, ax=axes, fraction=0.02, pad=0.01)
        cb.set_label(r"$\log_{10}(\mathrm{MSE})$")

    # small cleanup like your original
    axes[0, 0].set_yticklabels([])

    return fig, axes
```
