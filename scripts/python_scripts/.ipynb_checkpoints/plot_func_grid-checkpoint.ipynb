```python
import numpy as np
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
from matplotlib.offsetbox import AnchoredText
import matplotlib.patheffects as pe

def make_target_function(func_type: str):
    if func_type == "ReLU":
        m = nn.ReLU()
        return lambda x: m(x)
    if func_type == "Sigmoid":
        m = nn.Sigmoid()
        return lambda x: m(4 * x)
    if func_type == "Parabola":
        return lambda x: x**2
    if func_type == "Cubic":
        return lambda x: x**3
    if func_type == "Sine":
        return lambda x: torch.sin(3 * x)
    if func_type == "Cosine":
        return lambda x: torch.cos(3 * x)
    if func_type == "Linear":
        return lambda x: x
    if func_type == "Gaussian":
        return lambda x: torch.exp(-x**2 / 2.0)
    else:
        raise ValueError(f"Unknown func_type={func_type}")


def load_condsm_for_datatype(CondSM_cls, ckpt_path: str, device="cuda"):
    state_dict = torch.load(ckpt_path, weights_only=False, map_location=device)
    model = CondSM_cls(state_dict["layer_dims"], state_dict["dropout"], state_dict["batch_norm"]).to(device)
    model.load_state_dict(state_dict["model_state_dict"])
    model.eval()
    return model


def build_input_tensor_from_npz(
    npz: dict,
    n: int,
    in_idx: int,
    x_min: float = -1.5,
    x_max: float =  1.5,
    device="cuda",
):
    true_output_idx = 0
    true_input_idx = in_idx + 1  # matches your convention

    # npz['control'] appears to be one best control vector (shape (8,) or similar)
    # You delete output idx and true input idx to obtain the 6 controls.
    ctrl6 = np.delete(npz["control"], [true_output_idx, true_input_idx])
    controls = torch.from_numpy(ctrl6).to(device).float().expand(n, -1)

    # Insert x at the right place so that x corresponds to original true_input_idx
    c_left, c_right = controls[:, :in_idx], controls[:, in_idx:]
    x = torch.linspace(x_min, x_max, n, device=device).unsqueeze(1)
    x_full = torch.cat([c_left, x, c_right], dim=-1)  # shape (n, 7)
    return x.squeeze(1), x_full


def fit_affine_to_target(
    y_sm: torch.Tensor,
    y_tgt: torch.Tensor,
    eps: float = 1e-12,
    clamp_gain_nonneg: bool = False
):
    """
    Fit a,b minimizing ||a*y_sm + b - y_tgt||^2 (least squares with intercept).
    Returns: (a, b, y_aff)
    """
    if not torch.is_tensor(y_sm):
        y_sm = torch.as_tensor(y_sm)
    if not torch.is_tensor(y_tgt):
        y_tgt = torch.as_tensor(y_tgt)

    y = y_sm.float().flatten()
    t = y_tgt.float().flatten()

    y_mean = y.mean()
    t_mean = t.mean()

    y0 = y - y_mean
    t0 = t - t_mean

    var_y = (y0 * y0).mean()
    cov_yt = (y0 * t0).mean()

    a = cov_yt / (var_y + eps)
    if clamp_gain_nonneg:
        a = torch.clamp(a, min=0.0)

    b = t_mean - a * y_mean
    y_aff = a * y_sm + b
    return a, b, y_aff


def mse(a, b):
    """Mean squared error between two 1D arrays/tensors."""
    if not torch.is_tensor(a):
        a = torch.as_tensor(a)
    if not torch.is_tensor(b):
        b = torch.as_tensor(b)
    a = a.float().flatten()
    b = b.float().flatten()
    return torch.mean((a - b) ** 2)
    
def rmse(a, b):
    """Mean squared error between two 1D arrays/tensors."""
    if not torch.is_tensor(a):
        a = torch.as_tensor(a)
    if not torch.is_tensor(b):
        b = torch.as_tensor(b)
    a = a.float().flatten()
    b = b.float().flatten()
    return torch.sqrt(torch.mean((a - b) ** 2))

def nrmse(a, ref):
    if not torch.is_tensor(a):
        a = torch.as_tensor(a)
    if not torch.is_tensor(ref):
        ref = torch.as_tensor(ref)
    a = a.float().flatten()
    ref = ref.float().flatten()
    rmse = torch.sqrt(torch.mean((a - ref) ** 2))
    
    q25 = torch.quantile(ref, 0.25)
    q75 = torch.quantile(ref, 0.75)
    S = q75 - q25
    nrmse = rmse / S
    return nrmse

def mae(a, b):
    """Mean squared error between two 1D arrays/tensors."""
    if not torch.is_tensor(a):
        a = torch.as_tensor(a)
    if not torch.is_tensor(b):
        b = torch.as_tensor(b)
    a = a.float().flatten()
    b = b.float().flatten()
    return torch.mean(torch.abs((a - b)))

def pearson_corrcoef(x: torch.Tensor, y: torch.Tensor, eps: float = 1e-8) -> torch.Tensor:
    """
    Pearson correlation between two same-shaped tensors.
    Returns a scalar tensor.

    Works for any shape as long as x and y have the same number of elements
    (e.g., (N,), (N,1), (B,N) if you flatten first, etc.).
    """
    if x.numel() != y.numel():
        raise ValueError(f"x and y must have the same number of elements, got {x.numel()} vs {y.numel()}")

    x = x.reshape(-1)
    y = y.reshape(-1)

    x = x - x.mean()
    y = y - y.mean()

    # (x·y) / (||x|| * ||y||)
    denom = (x.norm(p=2) * y.norm(p=2)).clamp_min(eps)
    return (x @ y) / denom

def plot_panel_raw(ax, x, y_sm, I_out, I_err, title=None, show_xlabel=False, show_ylabel=False):
    ax.plot(x, y_sm, lw=1.0, label="SM")
    ax.fill_between(x, I_out - I_err, I_out + I_err, alpha=0.25, label="kMC band", color="green")
    ax.plot(x, I_out, lw=0.9, label="kMC", color="green")
    if show_xlabel:
        ax.set_xlabel(r"$V$ [V]")
    if show_ylabel:
        ax.set_ylabel(r"$I_{\mathrm{out}}$ [nA]")
    if title:
        ax.set_title(title)


def plot_panel_affine(ax, x, y_sm, y_tgt, title=None, show_xlabel=False, show_ylabel=False):
    a, b, y_sm_aff = fit_affine_to_target(y_sm, y_tgt)
    ax.plot(x, y_sm_aff, lw=1.0, label="SM")
    ax.plot(x, y_tgt, lw=1.0, label="Target")
    if show_xlabel:
        ax.set_xlabel(r"$V$ [V]")
    if show_ylabel:
        ax.set_ylabel("$I_{\mathrm{out}}$ [nA]")
    if title:
        ax.set_title(title)
    return a, b, y_sm_aff


"""def add_pretty_mse_box(ax, txt: str, loc: str = "upper left", fontsize: int = 7):
    at = AnchoredText(
        txt,
        loc=loc,
        prop=dict(size=fontsize),
        pad=0.25,
        borderpad=0.3,
        frameon=True,
    )
    at.patch.set_boxstyle("round,pad=0.25,rounding_size=0.2")
    at.patch.set_facecolor("white")
    at.patch.set_alpha(0.85)
    at.patch.set_edgecolor((0, 0, 0, 0.25))  # light border
    at.patch.set_linewidth(0.8)
    ax.add_artist(at)
"""
def add_pretty_mse_box(
    ax,
    txt: str,
    loc: str = "upper left",
    fontsize: int = 5,
):
    at = AnchoredText(
        txt,
        loc=loc,
        prop=dict(size=fontsize),
        pad=0.25,
        borderpad=0.35,
        frameon=True,
    )
    at.patch.set_facecolor("lightyellow")

    # Rounded rectangle with slightly more padding
    at.patch.set_boxstyle("round,pad=0.35,rounding_size=0.25")

    # "Card" styling
    at.patch.set_facecolor("white")
    at.patch.set_alpha(0.90)
    at.patch.set_edgecolor((0, 0, 0, 0.18))
    at.patch.set_linewidth(0.9)

    # Subtle shadow (really helps)
    at.patch.set_path_effects([
        pe.SimplePatchShadow(offset=(1.2, -1.2), alpha=0.18, rho=0.98),
        pe.Normal(),
    ])

    ax.add_artist(at)
    return at

def plot_inverse_design_grid_dual(
    CondSM_cls,
    model_ckpt_template: str,
    npz_template: str,
    out_pdf: str,
    func_types: list[str],
    data_types: list[str],
    in_idx: int,
    seed: int,
    n: int = 50,
    time_scale: float = 1e21,
    x_min: float = -1.5,
    x_max: float =  1.5,
    device: str = "cuda",
):
    """
    Creates a grid with 2 rows per function:
      For each function:
        row 2k: raw
        row 2k+1: z
      cols: data_types

    Each subplot shows ONE MSE: between the two curves actually plotted.
      - raw row: MSE(SM, kMC mean)
      - z row:   MSE(affine(SM), Target)
    """
    doubled_funcs = []
    for f in func_types:
        doubled_funcs.append((f, "raw"))
        doubled_funcs.append((f, "z"))

    n_rows = len(doubled_funcs)
    n_cols = len(data_types)

    models = {}
    for dt in data_types:
        ckpt_path = model_ckpt_template.format(data_type=dt)
        models[dt] = load_condsm_for_datatype(CondSM_cls, ckpt_path, device=device)

    panel_w, panel_h = 2.2, 1.6
    fig, axes = plt.subplots(
        n_rows, n_cols,
        figsize=(panel_w * n_cols, panel_h * n_rows),
        sharex=True, sharey=False, squeeze=False
    )
    fig.subplots_adjust(left=0.06, right=0.995, bottom=0.06, top=0.94, wspace=0.18, hspace=0.18)

    for i, (func_type, mode) in enumerate(doubled_funcs):
        tgt_fn = make_target_function(func_type)

        for j, dt in enumerate(data_types):
            ax = axes[i, j]
            npz_path = npz_template.format(func_type=func_type, data_type=dt, in_idx=in_idx, seed=seed)
            data = np.load(npz_path)

            x, x_full = build_input_tensor_from_npz(
                data, n=n, in_idx=in_idx, x_min=x_min, x_max=x_max, device=device
            )

            with torch.no_grad():
                y_sm = models[dt](x_full).detach().cpu().squeeze()

            I_out = torch.tensor(data["current"], dtype=torch.float32) * time_scale
            I_err = torch.tensor(data["sem"], dtype=torch.float32) * time_scale

            ax.minorticks_on()
            ax.tick_params(direction="in", which="both", top=True, right=True)
            ax.grid(which="major", alpha=0.25, lw=0.5)

            if j == 0:
                ax.text(
                    -0.28, 0.5, f"{func_type}", rotation=90,
                    va="center", ha="center", transform=ax.transAxes, fontsize=8
                )
            """if j != 0:
                ax.tick_params(labelleft=False)"""

            # --- plot + compute ONE MSE (only for curves shown) ---
            if mode == "raw":
                y_tgt = tgt_fn(x.cpu()).float().squeeze()
                plot_panel_raw(
                    ax,
                    x.cpu().numpy(), y_sm.cpu().numpy(),
                    I_out.cpu().numpy(), I_err.cpu().numpy(),
                    title=None, show_xlabel=(i == 7), show_ylabel=(j == 0)
                )
                #m = mse(y_sm, I_out).item()
                
                r = mae(y_sm, I_out)
                _, _, I_out_aff = fit_affine_to_target(I_out, y_tgt)
                g_0, p_0, y_sm_aff = fit_affine_to_target(y_sm, y_tgt)
                I_out_aff = g_0 * I_out + p_0
                #e_aff = mse(I_out_aff, y_tgt)
                e_aff = mse(I_out_aff, y_tgt)
                nrmse_score = nrmse(y_sm, I_out)
                txt = f"$\\mathcal{{E}}_{{\\mathrm{{aff}}}}$(kMC, T): {e_aff:.2e}\n"\
                      f"NRMSE(SM,kMC): {nrmse_score:.2e}"  
                add_pretty_mse_box(ax, txt, loc="upper left", fontsize=4)
                #ax.set_title(txt)

            else:
                y_tgt = tgt_fn(x.cpu()).float().squeeze()
                g_0, p_0, y_sm_aff = fit_affine_to_target(y_sm, y_tgt)
                _, _, I_out_aff = fit_affine_to_target(I_out, y_tgt)
                
                plot_panel_affine(
                    ax,
                    x.cpu().numpy(),
                    y_sm.cpu().numpy(),
                    y_tgt.cpu().numpy(),
                    title=None, show_xlabel=(i == 7), show_ylabel=(j == 0)
                )

                r_1 = mae(y_sm_aff, y_tgt).item()
                r_2 = mae(I_out_aff, y_tgt).item()
                e_aff = mse(y_sm_aff, y_tgt)
                txt = f"$\\mathcal{{E}}_{{\\mathrm{{aff}}}}$(SM, T): {e_aff:.2e}\n"\
                      f"$G_0 = {g_0:.2f}$ $P_0 = {p_0:.2f}$"
                      #f"$MAE$(kMC, T): {r_2:.2e} $[\\mathrm{{nA}}]$"

                add_pretty_mse_box(ax, txt, loc="upper left", fontsize=4)

            # --- do NOT touch tick labels (removed your labelleft/labelbottom edits) ---
            # If you still want fewer axis labels, you can remove x/y labels here,
            # but leaving ticks/labels untouched as requested.

    # --- collect legend entries from all axes ---
    handles, labels = [], []
    for ax in fig.axes:
        h, l = ax.get_legend_handles_labels()
        handles.extend(h)
        labels.extend(l)

    # de-duplicate while preserving order
    seen = set()
    legend_handles, legend_labels = [], []
    for h, lab in zip(handles, labels):
        if lab and lab not in seen:
            legend_handles.append(h)
            legend_labels.append(lab)
            seen.add(lab)

    if legend_handles and legend_labels:
        fig.legend(
            legend_handles, legend_labels,
            loc="upper center",
            ncol=min(4, len(legend_labels)),
            frameon=False,
            bbox_to_anchor=(0.5, 0.97),
        )
    fig.savefig(out_pdf, bbox_inches='tight')
    plt.close(fig)
    print(f"Saved: {out_pdf}")
```
