```python
import numpy as np
import torch

# Assumption: gate_inputs are ordered as [00, 01, 10, 11]
# For AND: only state 11 is HIGH (1), others are LOW (0)
HIGH_IDX = torch.tensor([idx for idx, val in enumerate(truth_outputs) if val == 1.0], device=device)          # 11
LOW_IDX  = torch.tensor([idx for idx, val in enumerate(truth_outputs) if val == 0.0], device=device)    # 00,01,10
print(HIGH_IDX)
print(LOW_IDX)

seed = 42
torch.manual_seed(seed)

num_samples = 5_000_000
theta_samples = torch.rand(size=(num_samples, 5), device=device) * (c_v_max - c_v_min) + c_v_min

# Noise in SAME units as model output (current)
sigma_val = 0.01
sigma = float((sigma_val * S_std).item() if torch.is_tensor(S_std) else sigma_val * S_std)

chunk_size = 1024

model = model.to(device)
model.eval()

scores = np.empty(num_samples, dtype=np.float32)

# ---------- Scoring options ----------
# Option A (recommended): "sigma-margin" score (fastest, very effective)
# margin = min(I_high) - max(I_low); for AND with one high state => I11 - max(I00,I01,I10)
# score = margin / sigma
USE_SIGMA_MARGIN = True

# Option B: analytic noisy classification score using best threshold T per theta (no Monte Carlo)
# score = mean log P(correct) over the 4 states (stable, directly noise-robust)
USE_LOGPROB = False  # set True to use this instead of sigma-margin

eps = 1e-12

with torch.no_grad():
    for start in range(0, num_samples, chunk_size):
        end = min(start + chunk_size, num_samples)
        B = end - start

        theta_b = theta_samples[start:end].to(dtype=torch.float32)

        # Build input tensor for all (theta, gate_state):
        # input_b: [B, 4, num_voltages]
        input_b = torch.zeros((B, len(gate_inputs), num_voltages), device=device, dtype=torch.float32)

        # Fill control voltages
        for j, c_idx in enumerate(control_indices):
            input_b[:, :, c_idx] = theta_b[:, j].unsqueeze(1)  # [B,1] -> broadcast over 4 states

        # Fill the two gate input voltages for the 4 states
        gate_in = torch.tensor(gate_inputs, device=device, dtype=torch.float32)  # [4,2]
        input_b[:, :, input_index_0] = gate_in[:, 0].unsqueeze(0)  # [1,4]
        input_b[:, :, input_index_1] = gate_in[:, 1].unsqueeze(0)  # [1,4]

        # Forward pass: [B*4, num_voltages] -> [B,4] currents
        y0 = model(input_b.reshape(B * 4, num_voltages)).squeeze()
        y0 = y0.reshape(B, 4)  # [B,4]

        # ---- AND gate boolean scoring in current space ----
        I_high = y0.index_select(dim=1, index=HIGH_IDX)  # [B,1]
        I_low  = y0.index_select(dim=1, index=LOW_IDX)   # [B,3]

        # worst-case separation
        min_high = I_high.min(dim=1).values              # [B]
        max_low  = I_low.max(dim=1).values               # [B]
        margin   = min_high - max_low                    # [B]

        if USE_SIGMA_MARGIN:
            # Higher is better; margin>0 means perfectly separable by a single threshold
            robust = margin / (sigma + 1e-30)            # [B]

        elif USE_LOGPROB:
            # Best "separating" threshold halfway between worst low and worst high
            T = 0.5 * (min_high + max_low)               # [B]

            # Gaussian CDF Phi(x) = P(Z <= x), Z~N(0,1)
            # In PyTorch: torch.special.ndtr
            # P(I + eps > T) = 1 - Phi((T - I)/sigma)
            # P(I + eps <= T) = Phi((T - I)/sigma)
            z = (T.unsqueeze(1) - y0) / (sigma + 1e-30)  # [B,4]
            Phi = torch.special.ndtr(z)                  # [B,4]

            # Targets for AND (00,01,10,11) => [0,0,0,1]
            t = torch.tensor(truth_outputs, device=device, dtype=torch.float32).unsqueeze(0)  # [1,4]

            P_correct = torch.where(t > 0.5, 1.0 - Phi, Phi).clamp_min(eps)  # [B,4]
            robust = torch.log(P_correct).mean(dim=1)     # [B]

        else:
            raise ValueError("Select a scoring option.")

        scores[start:end] = robust.detach().cpu().numpy().astype(np.float32)

print("Done. scores shape:", scores.shape)
best_idx = int(np.argmax(scores))
best_theta = theta_samples[best_idx].detach().cpu()
best_score = float(scores[best_idx])
print("Best idx:", best_idx, "best_score:", best_score)
print("Best theta:", best_theta.tolist())

np.savetxt(
    f'/gpfs/bwfor/work/ws/hd_gy283-my_data_new/logic_gates_data/{data_type}_gate_data/'
    f'{data_type}_gate={gate}_params_seed={seed}_sigma_val={sigma_val}_score={"sigmamargin" if USE_SIGMA_MARGIN else "logprob"}.txt',
    best_theta.tolist()
)
```
