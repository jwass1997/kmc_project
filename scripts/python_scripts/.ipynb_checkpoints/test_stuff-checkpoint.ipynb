## vMB 

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches

from mpl_toolkits.axes_grid1.anchored_artists import AnchoredSizeBar

N = 200
R = 150
alpha = 0.5
beta = 5
mu = 0
kappa = 2

def Beta_von_Mises_component(alpha,
                             beta,
                             mu,
                             kappa,
                             radius,
                             N):
    
    u = np.random.beta(a=alpha, b=beta, size=N)

    r = radius*np.sqrt(u)

    theta = np.random.vonmises(mu=mu, kappa=kappa, size=N)

    sample = np.column_stack([r*np.cos(theta), r*np.sin(theta)])
    
    return sample

def sample_from_Beta_von_Mises_mixture(alphas,
                                       betas,
                                       mus,
                                       kappas,
                                       weights,
                                       radius,
                                       N):
    
    cum_weight_sum = np.cumsum(weights)
    sample = []

    for _ in range(N):

        u = np.random.uniform(0., 1.)*cum_weight_sum[-1]
        component = np.searchsorted(cum_weight_sum, u)

        x = Beta_von_Mises_component(
            alpha=alphas[component],
            beta=betas[component],
            mu=mus[component],
            kappa=kappas[component],
            radius=radius,
            N=1
        )

        sample.append(x)

    return np.vstack(sample)

num_of_components = 1

alphas = 1
betas = 1
mus = 2*np.pi
kappas = 0.1

raw_weights = np.random.uniform(0, 1, num_of_components)
cum_weight_sum = np.cumsum(raw_weights)
weights = [w / cum_weight_sum[-1] for w in raw_weights]

positions = Beta_von_Mises_component(
    alphas,
    betas,
    mus,
    kappas,
    R,
    N
)

with open("/home/hd/hd_hd/hd_gy283/kmc_project/configs/vmb_test.txt", "w") as f:
    for x in positions:
        f.write(f"{x[0]}\t{x[1]}\n")

scatter_args = {
    "s": 15.,
    "alpha": .7,
}

fig, ax = plt.subplots(figsize=(6, 6))
ax.scatter(*positions.T, **scatter_args)
boundary = patches.Circle(xy=(0, 0), radius=R, fill=False, edgecolor="grey", lw=4)
ax.add_patch(boundary)
ax.axis('off')

scalebar = AnchoredSizeBar(
    ax.transData,
    size=60,
    label="$d_{\mathrm{max}}$",
    loc="upper right",
    pad=0.1, borderpad=1.5, sep=2, size_vertical=5,
    frameon=False,
    fill_bar=True
)

ax.add_artist(scalebar)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches

from mpl_toolkits.axes_grid1.anchored_artists import AnchoredSizeBar

N = 200
R = 150
alpha = 0.5
beta = 5
mu = 0
kappa = 2

def Beta_von_Mises_component(alpha,
                             beta,
                             mu,
                             kappa,
                             radius,
                             N):
    
    u = np.random.beta(a=alpha, b=beta, size=N)
    r = radius*np.sqrt(u)

    theta = np.random.vonmises(mu=mu, kappa=kappa, size=N)

    sample = np.column_stack([r*np.cos(theta), r*np.sin(theta)])
    
    return sample

def sample_from_Beta_von_Mises_mixture(alphas,
                                       betas,
                                       mus,
                                       kappas,
                                       weights,
                                       radius,
                                       N):
    
    cum_weight_sum = np.cumsum(weights)
    sample = []

    for _ in range(N):

        u = np.random.uniform(0., 1.)*cum_weight_sum[-1]
        component = np.searchsorted(cum_weight_sum, u)

        x = Beta_von_Mises_component(
            alpha=alphas[component],
            beta=betas[component],
            mu=mus[component],
            kappa=kappas[component],
            radius=radius,
            N=1
        )

        sample.append(x)

    return np.vstack(sample)

num_of_components = 1

alphas = 4.5*np.random.uniform(0, 1, N) + 0.5
betas = 4.5*np.random.uniform(0, 1, N) + 0.5
mus = 2*np.pi*np.random.uniform(0, 1, N)
kappas = 0.9*np.random.uniform(0, 1, N)

raw_weights = np.random.uniform(0, 1, num_of_components)
cum_weight_sum = np.cumsum(raw_weights)
weights = [w / cum_weight_sum[-1] for w in raw_weights]

positions = sample_from_Beta_von_Mises_mixture(
    alphas,
    betas,
    mus,
    kappas,
    weights,
    R,
    N
)

with open("../../configs/vonMises_beta_SM.txt", "w") as f:
    for x in positions:
        f.write(f"{x[0]}\t{x[1]}\n")

scatter_args = {
    "s": 15.,
    "alpha": .7,
}

fig, ax = plt.subplots(figsize=(6, 6))
ax.scatter(*positions.T, **scatter_args)
boundary = patches.Circle(xy=(0, 0), radius=R, fill=False, edgecolor="grey", lw=4)
ax.add_patch(boundary)
ax.axis('off')

scalebar = AnchoredSizeBar(
    ax.transData,
    size=60,
    label="$d_{\mathrm{max}}$",
    loc="upper right",
    pad=0.1, borderpad=1.5, sep=2, size_vertical=5,
    frameon=False,
    fill_bar=True
)

ax.add_artist(scalebar)
```

```python
def sample(means,
           covs,
           weights,
           radius,
           N):
    
    positions = []
    cum_sum_weights = np.cumsum(weights)

    while len(positions) < N:

        u = np.random.uniform(0.0, 1.0) * cum_sum_weights[-1]
        component = np.searchsorted(cum_sum_weights, u)

        x = np.random.multivariate_normal(mean=means[component], cov=covs[component])

        if np.sqrt(x[0]**2 + x[1]**2) < radius:

            positions.append(x) 
    
    return np.vstack(positions)

K = 5
radius = 150
n_acc = 200

R = np.sqrt(np.pi*radius**2 / n_acc)
radius_normalized = radius / R

std_min = radius / 3
std_max = radius / 2

log_bound_min = 2*np.log(std_min)
log_bound_max = 2*np.log(std_max)

means = []
covs = []
weights = []

raw_weights = []
for k in range(K):
    u = np.random.random(size=(1,))
    r = radius*np.sqrt(u)

    theta = 2*np.pi*np.random.random((1,))
    means.append(np.array([r*np.cos(theta), r*np.sin(theta)]).squeeze(-1))
    covs.append(np.exp((log_bound_max - log_bound_min)*np.random.random_sample() + log_bound_min) * np.eye(2))
    raw_weight = np.random.random()
    raw_weights.append(raw_weight)

raw_weight_sum = sum(raw_weights)
weights = [rw / raw_weight_sum for rw in raw_weights]

positions = sample(means, covs, weights, radius, 200)

scatter_args = {
    "s": 15.,
    "alpha": .7,
}

fig, ax = plt.subplots(figsize=(6, 6))
ax.scatter(*positions.T, **scatter_args)
boundary = patches.Circle(xy=(0, 0), radius=radius, fill=False, edgecolor="grey", lw=4)
ax.add_patch(boundary)
ax.axis('off')

scalebar = AnchoredSizeBar(
    ax.transData,
    size=60,
    label="max_dist",
    loc="upper right",
    pad=0.1, borderpad=1.5, sep=2, size_vertical=2.5,
    frameon=False
)

ax.add_artist(scalebar)
```

```python
def create_2d_mixture_of_truncated_gaussians(means: np.array, 
                                   covariance_matrices: np.array, 
                                   weights: np.array, 
                                   radius: float,
                                   N: int) -> np.array:
    
    cum_sum_weights = np.cumsum(weights)

    positions = []

    while len(positions) < N:

        u = np.random.uniform(0, 1) * cum_sum_weights[-1]
        component = np.searchsorted(cum_sum_weights, u)

        x = np.random.multivariate_normal(mean=means[component], cov=covariance_matrices[component])

        if np.sqrt(x[0]**2 + x[1]**2) < radius:

            positions.append(x)
    
    return positions
```

```python
means = np.array([[-0.4, -0.6], [0.2, 0.4], [0.5, -0.5]])
covariance_matrices = np.array([0.02*np.eye(2), 0.05*np.eye(2), 0.04*np.eye(2)])
weights = np.array([0.3, 0.3, 0.4])

pos = create_2d_mixture_of_truncated_gaussians(
    means==means,
    covariance_matrices=covariance_matrices,
    weights=weights,
    radius=1,
    N=10_000
)
pos_arr = np.array(pos)

plt.figure(figsize=(5, 5))
plt.scatter(pos_arr[:, 0], pos_arr[:, 1], s=2.5)
```

```python
import torch
import torch.nn as nn

from torch.utils.data import DataLoader
from sklearn.model_selection import train_test_split
from train_sm import MakeDataset
from sm import NeuralNet

state_dict = torch.load("/home/hd/hd_hd/hd_gy283/kmc_project/models/sm_vMB_1e7.pth", map_location="cpu", weights_only=False)
args = state_dict['args']
#print(args.keys())
model = NeuralNet(args['layer_dims'])
#print(model)
#print(model.layer_dims[0])

class LearnableActivationFunc(nn.Module):
    def __init__(self, model, v_min, v_max, device):
        super().__init__()

        self.v_min = v_min
        self.v_max = v_max

        self.sm = copy.deepcopy(model)

        self.device = device

        if device is not None:
            self.sm = self.sm.to(device)

        for p in self.sm.parameters():
            if p.requires_grad == True:
                p.requires_grad = False
        self.sm.eval()
        
        self.reset_cntrl_parameters()

    def reset_cntrl_parameters(self):

        num_control_parameters = self.sm.layer_dims[0] - 1

        with torch.no_grad():
            #init_params = (self.v_max - self.v_min) * torch.randn(num_control_parameters, device=self.device)
            init_params = torch.tensor([0.6851745748511203, 0.794845834384324, -0.6914577581682977, -1.5, -1.5, 1.5], dtype=torch.float32).to(self.device)

        self.raw_cntrl_params = nn.Parameter(init_params)

    def scale_cntrl(self):

        return self.v_min + (self.v_max - self.v_min) * (torch.tanh(self.raw_cntrl_params) + 1) / 2

    def affine_input_scaling(self, x, eps=1e-8):

        x_min = x.min()
        x_max = x.max()

        scale = (self.v_max - self.v_min) / (x_max - x_min + eps)
        
        return self.v_min + (x - x_min) * scale 

    def forward(self, x):

        orig_shape = x.shape

        x = x.reshape(-1, 1)
        B = x.size(0)

        cntrl_params = self.scale_cntrl().to(x.device, x.dtype)[None,:].expand(B, -1)
        x_sc = self.affine_input_scaling(x)
        input_tensor = torch.cat([x_sc, cntrl_params], dim=1)

        out = self.sm(input_tensor)

        return out.reshape(orig_shape) 

class ComposedFunction(nn.Module):

    def __init__(self, state_dict, num_basis_components, in_min, in_max, in_features, out_features, input_idx, device):
        super().__init__()

        self.in_features = in_features
        self.out_features = out_features
        self.input_idx = input_idx
        self.device =  device

        self.in_min = in_min
        self.in_max = in_max
        
        self.num_basis_components = num_basis_components
        
        self.basis_functions = self.build_basis_functions(state_dict)

        self.init_control_parameters()

        self.output_layer = nn.Linear(self.num_basis_components, self.out_features)

        self.to(self.device)

    def build_basis_functions(self, state_dict):
        
        module_list = nn.ModuleList()
        
        for _ in range(self.num_basis_components):
            
            net_i = NeuralNet(state_dict['args']['layer_dims'])
            net_i.load_state_dict(state_dict['model_state_dict'])
            
            net_i.eval()
            for p in net_i.parameters():
                p.requires_grad = False
            module_list.append(net_i)

        return module_list

    def init_control_parameters(self):

        num_control_params = self.in_features - 1
        
        with torch.no_grad():
            init_params = (self.in_max - self.in_min) * torch.rand(self.num_basis_components, num_control_params) + self.in_min           
            
        self.control_parameters = nn.Parameter(init_params)

    def forward(self, x):

        param_dtype = next(self.parameters()).dtype
        x = x.to(self.device, dtype=param_dtype)
        B = x.size(0)
        
        if x.dim() == 1:
            x = x.unsqueeze(1)
            
        y = x.new_empty(B, self.num_basis_components)
        
        for idx, basis_func in enumerate(self.basis_functions):
            cntrl_i = self.control_parameters[idx].to(self.device).unsqueeze(0).expand(B, -1)
            inp_i = torch.cat([x, cntrl_i], dim=-1)
            out_i = basis_func(inp_i)
            y[:, idx] = out_i.view(B)

        out = self.output_layer(y)
            
        return out
```

```python
comp_model = ComposedFunction(state_dict=state_dict, num_basis_components=10, in_min=-1.5, in_max=1.5, in_features=7, out_features=1, input_idx=0, device=device)
trainable_params = [comp_model.control_parameters] + list(comp_model.output_layer.parameters())
optimizer = torch.optim.Adam(trainable_params, lr=1e-3)

mse_loss = nn.MSELoss()

class ZScoreMSELoss(nn.Module):
    def __init__(self, eps=1e-8, dim=0, unbiased=False):
        super().__init__()
        self.eps = eps
        self.dim = dim
        self.unbiased = unbiased
        self.mse = nn.MSELoss()

    def zscore(self, x):
        # flatten across batch/features you want to normalize over
        mean = x.mean(dim=self.dim, keepdim=True)
        std  = x.std(dim=self.dim, unbiased=self.unbiased, keepdim=True).clamp_min(self.eps)
        return (x - mean) / std

    def forward(self, pred, target):
        # For shapes [B,1], set dim=0 (normalize across batch)
        zp = self.zscore(pred)
        zt = self.zscore(target)
        return self.mse(zp, zt)

own_loss = ZScoreMSELoss()
        
x = np.linspace(-1.5, 1.5, 1000)

def target_function(x):

    return 1 / (1 + np.exp(-5*x))

y = target_function(x)

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

train_data = MakeDataset(x_train, y_train)
test_data = MakeDataset(x_test, y_test)

batch_size = 800
train_loader = DataLoader(train_data, batch_size=batch_size, shuffle=True)
test_loader = DataLoader(test_data, batch_size=batch_size, shuffle=False)

epochs = 1000

for epoch in range(1, epochs):
    running_train_loss = 0.0    
    comp_model.train()
    for x, y in train_loader:
        x, y = x.to(device), y.to(device)

        optimizer.zero_grad()

        pred = comp_model(x)

        loss = own_loss(pred, y)    
        loss.backward()
        optimizer.step()
        with torch.no_grad():
            comp_model.control_parameters.clamp_(comp_model.in_min, comp_model.in_max)

        running_train_loss += loss.item() * x.size(0)
    epoch_train_loss = running_train_loss / len(train_loader.dataset)

    comp_model.eval()
    running_test_loss = 0.0
    with torch.no_grad():
        for x, y in test_loader:
            x, y = x.to(device), y.to(device)
            pred = comp_model(x)

            loss = own_loss(pred, y)

            running_test_loss += loss.item() * x.size(0)

        epoch_test_loss = running_test_loss / len(test_loader.dataset)

    if epoch % 100 == 0:
        print(f"[EPOCH] {epoch}/{epochs} | train_loss: {epoch_train_loss:.6f} | test_loss: {epoch_test_loss:.6f}\n")
```
